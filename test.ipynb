```python
import h5py
import torch
import torch.nn.functional as F
```

```python
dataset_path = '/users/zzeng28/data/zzeng28/repo/lardp/data/robomimic/tool_hang/ph/image_v15_prep_256_abs.hdf5'
# dataset_path = 'data/robomimic/tool_hang/ph/low_dim_v15.hdf5'
camera_view =  'sideview_image' # 'robot0_eye_in_hand_image'  # sideview_image


with h5py.File(dataset_path) as file:
    # count total steps
    print(file["data"].attrs['env_args'])
    demo = file["data"]["demo_1"]
    print(demo.keys())
    actions = demo["actions"][()]
    dino_patch_mean = demo["dino_patch_mean"]["vits16plus"][camera_view][()]
    dino_cls = demo["dino_cls"]["vits16plus"][camera_view][()]
    la_fs1 = demo["latent_actions"]["fs1"][camera_view][()]
    la_fs10 = demo["latent_actions"]["fs10"][camera_view][()]
    la_prebn_fs1 = demo["latent_actions_prebn"]["fs1"][camera_view][()]
    la_prebn_fs10 = demo["latent_actions_prebn"]["fs10"][camera_view][()]
```

```python
norm_dino_patch_mean = F.normalize(torch.from_numpy(dino_patch_mean), dim=-1)
norm_dino_cls = F.normalize(torch.from_numpy(dino_cls), dim=-1)

neg_cos_sim_patch_mean = -(norm_dino_patch_mean[0] * norm_dino_patch_mean[-1]).sum(dim=-1)
neg_cos_sim_cls = -(norm_dino_cls[0] * norm_dino_cls[-1]).sum(dim=-1)
print(neg_cos_sim_patch_mean)
print(neg_cos_sim_cls)
```

```python
norm_la_fs1 = F.normalize(torch.from_numpy(la_fs1), dim=-1)
norm_la_fs10 = F.normalize(torch.from_numpy(la_fs10), dim=-1)

neg_cos_sim_la_fs1 = -(norm_la_fs1[0] * norm_la_fs1[-1]).sum(dim=-1)
neg_cos_sim_la_fs10 = -(norm_la_fs10[0] * norm_la_fs10[-1]).sum(dim=-1)
print(neg_cos_sim_la_fs1)
print(neg_cos_sim_la_fs10)
```

```python
norm_la_prebn_fs1 = F.normalize(torch.from_numpy(la_prebn_fs1), dim=-1)
norm_la_prebn_fs10 = F.normalize(torch.from_numpy(la_prebn_fs10), dim=-1)

neg_cos_sim_la_prebn_fs1 = -(norm_la_prebn_fs1[0] * norm_la_prebn_fs1[-1]).sum(dim=-1)
neg_cos_sim_la_prebn_fs10 = -(norm_la_prebn_fs10[0] * norm_la_prebn_fs10[-1]).sum(dim=-1)
print(neg_cos_sim_la_prebn_fs1)
print(neg_cos_sim_la_prebn_fs10)
```
